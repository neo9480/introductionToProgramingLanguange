## Practice 2

```python
# • Create a random integer array A with shape (3,4) and values 0–9.
A = np.random.randint(0, 10, size=(3,4))
print(A)

# • Compute A + 5 and A * 2 without loops.
B = A + 2
C = A * 2
print(B, C)

# • Compute total, overall mean, mean of each row and maximum of each column.
total = A.sum()
mean = A.mean()
row_mean = A.mean(axis=1)
col_max = A.max(axis=0)

print(f"total = {total}, mean = {mean}, row_mean = {row_mean}, col_max = {col_max}")

# • Create a length-4 adjustment vector and add it to every row of A using broadcasting.
adjustment = np.array([1, 2, 3, 4])

D = A + adjustment

print("Adjustment vector:", adjustment)
print("A + adjustment: ", D)

# • Create a compatible 4 × 4 array B and compute A @ B.
B = np.random.randint(0, 10, size=(4, 4))

result = A @ B

print("B: ", B)
print("A @ B: ", result)
```

## Practice 3

```python
# • Create a 4 × 5 random integer array A with values 0–9.
A = np.random.randint(0, 10, size=(4, 5))
print(A)

# • Retrieve A[2,3], then the first 3 rows and first 4 columns.
print(f"A[2,3]: {A[2, 3]}")
print(f"first three rows: {A[:3, :4]}")
print(f"fist two columns: {A[::2, ::2]}")

# • Retrieve every other row and every other column.
print(f"every row and col: {A[:]}")

# • Set A[1,2] to 99.
A[1,2] = 99
print(f"new A: {A}")
# • Select all values greater than 5 using a Boolean mask.
mask = A > 5
print(f"values greater than 5: {A[mask]}")
# • Compute the sum of each row without a Python loop.
sum = A.sum(axis=1)
print(f"sum of rows: {sum}")
```

```python
# A slice is a view: changing it also changes A.
block_view = A[:2, :2]
block_view[0, 0] = 100
print("After modifying the view: ", A)

# .copy() makes an independent array: changing it leaves A unchanged.
block_copy = A[:2, :2].copy()
block_copy[0, 0] = 200
print("Modified copy: ", block_copy)
print("A remains unchanged by the copy modification: ", A)
```

## Long practice

### Part A

```python
# data
units = np.array(
    [
        [12, 15, 18, 14],  # Product A
        [20, 22, 19, 25],  # Product B
        [8, 10, 12, 11],  # Product C
        [16, 14, 17, 20],  # Product D
    ]
)
prices = np.array([25.0, 18.0, 40.0, 30.0])
```

```python
# • 1. Display ndim, shape, size and dtype for units and prices.
print(f"Units:: ndim: {units.ndim}, shape: {units.shape}, size: {units.size}, dtype: {units.dtype}")
print(f"Prices:: ndim: {prices.ndim}, shape: {prices.shape}, size: {prices.size}, dtype: {prices.dtype}")

# • 2. Compute total units sold for each product across all four weeks.
total_row = units.sum(axis=1)
print(f"total units sold for each product across all four weeks: {total_row}")

# • 3. Compute total units sold in each week across all products.
total_col = units.sum(axis=0)
print(f"total units sold in each week across all products: {total_col}")

# • 4. Identify the maximum and mean weekly unit sales.
max_sales = units.max(axis=0)
mean_sales = units.mean(axis=0)
print(f"Max sales: {max_sales}, mean sales: {mean_sales}")

# • 5. Retrieve the sales for Product B in Week 4 and the first two weeks for all products.
print("Product B, Week 4:", units[1, 3])
print("First two weeks for all products: ", units[:, :2])
```

### Part B

```python
# • 6. Reshape prices if necessary so each product price multiplies its full row of unit sales.
prices_by_product = prices[:, np.newaxis]

# • 7. Create a revenue array with the same shape as units.
revenue = units * prices_by_product
print("Revenue by product and week:\n", revenue)

# • 8. Compute total revenue per product and total revenue per week.
revenue_per_product = revenue.sum(axis=1)
revenue_per_week = revenue.sum(axis=0)
print("Total revenue per product:", revenue_per_product)
print("Total revenue per week:", revenue_per_week)

# • 9. Find all revenue cells above €500 using a Boolean mask.
high_revenue_mask = revenue > 500
print("Revenue cells above €500:\n", revenue[high_revenue_mask])

# • 10. Apply a 10% price increase to all products using vectorized arithmetic and recompute revenue.
increased_prices = prices * 1.10
increased_revenue = units * increased_prices[:, np.newaxis]
print("Revenue after 10% price increase:\n", increased_revenue)
```
