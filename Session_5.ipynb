## Practice 1

```python
import numpy as np
```

```python
# Create a 1D array containing multiples of 3 from 0 up to 99.
arr = np.arange(0, 100, 3)
print(arr)

#  Inspect ndim, size, shape and dtype.
print(arr.ndim)
print(arr.dtype)
print(arr.size)
print(arr.shape)

# Reshape it into a 2 × 17 array
Arr = arr.reshape(2, 17)
print(Arr)

# Create: zeros with shape (2,3,4), random integers with shape (2,4), and a 4 × 4 identity matrix.
zeros_array = np.zeros((2, 3, 4))
random_integers = np.random.randint(0, 100, size=(2, 4))
identity_matrix = np.eye(4)

print(zeros_array)
print(random_integers)
print(identity_matrix)

# Compare the memory used by a float64 version and a float32 version of the same 34-value array using .nbytes.
arr_float64 = arr.astype(np.float64)
arr_float32 = arr.astype(np.float32)

print(f"float64: {arr_float64.nbytes} bytes")
print(f"float32: {arr_float32.nbytes} bytes")
```

## Practice 2

```python
# • Create a random integer array A with shape (3,4) and values 0–9.
A = np.random.randint(0, 10, size=(3,4))
print(A)

# • Compute A + 5 and A * 2 without loops.
B = A + 2
C = A * 2
print(B, C)

# • Compute total, overall mean, mean of each row and maximum of each column.
total = A.sum()
mean = A.mean()
row_mean = A.mean(axis=1)
col_max = A.max(axis=0)

print(f"total = {total}, mean = {mean}, row_mean = {row_mean}, col_max = {col_max}")

# • Create a length-4 adjustment vector and add it to every row of A using broadcasting.
adjustment = np.array([1, 2, 3, 4])

D = A + adjustment

print("Adjustment vector:", adjustment)
print("A + adjustment:\n", D)

# • Create a compatible 4 × 4 array B and compute A @ B.
B = np.random.randint(0, 10, size=(4, 4))

result = A @ B

print("B:\n", B)
print("A @ B:\n", result)
```

## Practice 3

```python
# • Create a 4 × 5 random integer array A with values 0–9.
A = np.random.randint(0, 10, size=(4, 5))
print(A)

# • Retrieve A[2,3], then the first 3 rows and first 4 columns.
print(f"A[2,3]: {A[2, 3]}")
print(f"first three rows: {A[:3, :4]}")
print(f"fist two columns: {A[::2, ::2]}")

# • Retrieve every other row and every other column.
print(f"every row and col: {A[:]}")

# • Set A[1,2] to 99.
A[1,2] = 99
print(f"new A: {A}")
# • Select all values greater than 5 using a Boolean mask.
mask = A > 5
print(f"values greater than 5: {A[mask]}")
# • Compute the sum of each row without a Python loop.
sum = A.sum(axis=1)
print(f"sum of rows: {sum}")
```
