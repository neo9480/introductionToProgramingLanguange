```python
quantity = 12
unit_price = 49.90
discount_rate = 0.10
cost_per_unit = 31.00
```

```python
gross_revenue = unit_price * quantity
discount_value = gross_revenue * discount_rate
net_revenue = gross_revenue - discount_value
profit = net_revenue - (cost_per_unit * quantity)
profit_margin = (profit / net_revenue) * 100
```

```python
print(gross_revenue)
print(discount_value)
print(net_revenue)
print(profit)
print("%", profit_margin)
```
