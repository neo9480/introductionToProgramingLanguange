## Practice 2

```python
order_amount = int(input("Enter your order amount: "))
student = input("Are you a student 'yes' or 'no': ")


if order_amount >= 100:
  discount_rate = 10
elif 100 > order_amount >= 50:
  discount_rate = 5
else:
  discount_rate = 0

discounted_total = order_amount - ((discount_rate/100) * order_amount)

if student == "yes":
    discounted_total = discounted_total * 0.98
    
print(
    f"Order amount: {order_amount} \nThis is your discounted total: {discounted_total}"
)
```

## practice 5

```python
products = [
    {"name": "Book", "price": 12, "stock": 30},
    {"name": "Pen", "price": 2, "stock": 200},
    {"name": "Bag", "price": 35, "stock": 8},
]

# 1. Print each product name and price
for p in products:
    print(f"{p['name']}: ${p['price']}")

# 2. Total stock value of each product (price x stock)
print()
for p in products:
    value = p["price"] * p["stock"]
    print(f"{p['name']} stock value: ${value}")

# 3. Only products with stock below 20
print()
for p in products:
    if p["stock"] < 20:
        print(f"Low stock: {p['name']} ({p['stock']} left)")

# 4. list comprehension for expensive products (price > 15)
expensive = []
for p in products:
    if p["price"] > 15:
        expensive.append(p["name"])
print(f"Expensive Products: {expensive}")
```
