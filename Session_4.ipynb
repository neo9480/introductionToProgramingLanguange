```python
def make_shirt(size="L", message="I love Python"):
    return f"The shirt size is {size} and the message is '{message}'."


# Positional arguments
print(make_shirt("M", "Keep Learning"))

# Keyword arguments
print(make_shirt(message="Hello World", size="S"))

# Default arguments
print(make_shirt())


def describe_city(city, country="France"):
    return f"{city} is in {country}."


# Three cities, including one outside France
city1 = describe_city("Rennes")
city2 = describe_city("Paris")
city3 = describe_city("Tokyo", "Japan")

# Build a list of returned descriptions
cities = [city1, city2, city3]

for city in cities:
    print(city)
```

## PRACTICE 3

```python
def calculate_subtotal(price, quantity):
    return price * quantity

def apply_discount(amount, discount_rate=0.0):
    return amount * (1 - discount_rate)

def add_vat(amount, vat_rate=0.20):
    return amount * (1 + vat_rate)

def final_price(price, quantity, discount_rate=0.0):
    return add_vat(apply_discount(calculate_subtotal(price, quantity), discount_rate))

# Test three orders
order1 = final_price(10, 2)
order2 = final_price(25, 3)
order3 = final_price(50, 2, 0.10)

print(f"Order 1 total: €{order1:.2f}")
print(f"Order 2 total: €{order2:.2f}")
print(f"Order 3 total: €{order3:.2f}")
```

## PRACTICE 4

```python
class Book:
    def __init__(self, title, author, num_pages, category="General"):
        self.title = title
        self.author = author
        self.num_pages = num_pages
        self.category = category

    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")

    def is_long(self):
        return self.num_pages > 300

    def reading_time(self, minutes_per_page=2):
        return self.num_pages * minutes_per_page
```

```python
# Create two Book objects
book1 = Book("Harry Potter", "J.K. Rowling", 350, "Fantasy")
book2 = Book("The Alchemist", "Paulo Coelho", 208)
```

```python
# Test the first book
book1.display_info()
print("Is it a long book?", book1.is_long())
print("Category:", book1.category)
print("Estimated reading time:", book1.reading_time(), "minutes")

print()
```

```python
# Test the second book
book2.display_info()
print("Is it a long book?", book2.is_long())
print("Category:", book2.category)
print("Estimated reading time:", book2.reading_time(), "minutes")
```

```python
def order_subtotal(order):
    return order["price"] * order["qty"]


def apply_discount(amount, discount_rate=0.0):
    return amount * (1 - discount_rate)


def add_vat(amount, vat_rate=0.20):
    return amount * (1 + vat_rate)


def order_total(order, discount_rate=0.0):
    subtotal = order_subtotal(order)
    discounted_amount = apply_discount(subtotal, discount_rate)
    total = add_vat(discounted_amount)
    return total


# Test with one order
order = {"price": 50, "qty": 4}

print(order_subtotal(order))
print(apply_discount(200, 0.10))
print(add_vat(200))
print(order_total(order, 0.10))
```
