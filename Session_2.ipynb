```python
products = ["Book", "Pen", "Bag"]
prices = [12, 2, 35]

print("-----Order Summary-----")

total = 0
for product, price in zip(products, prices):
  print(f"{product}: ${price}")
  total += price

print(f"total: {total}")
print("-----------------------")
```
